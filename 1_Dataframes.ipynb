# Dataframes

Un DataFrame es una estructura de datos bidimensional, similar a una hoja de cálculo o una tabla SQL, que se utiliza en lenguajes como Python (con la librería Pandas) y R. Permite organizar datos en filas y columnas, donde cada columna puede contener datos de un tipo diferente (enteros, cadenas, booleanos, etc.)

```
# Crear un DataFrame desde un diccionario
import pandas as pd


datos = {
    'Nombre': ['Ana', 'Luis', 'Pedro'],
    'Edad': [23, 34, 29],
    'Ciudad': ['Bogotá', 'Medellín', 'Cali']
}

df = pd.DataFrame(datos)
```

```
#   df.select_dtypes(include='object')
#   select_dtypes() es una función de pandas que te permite filtrar columnas por su tipo de dato.
#include='object' indica que se deben incluir solo las columnas de tipo "object", que normalmente son cadenas de texto (strings).

lista= df.select_dtypes(include='object')

lista2=lista.columns
```

```
# Veamos la descripcion de las variables

for col in lista2:
    print(f"Frecuencia de valores en '{col}':")
    print(df[col].value_counts())
    print()
```

```

# Existe una forma mas compacta

import pandas as pd

# Suponiendo que ya cargaste el DataFrame como df
resumen = []

for col in df.columns:
    tipo = df[col].dtype
    nulos = df[col].isnull().sum()  #
    unicos = df[col].nunique()

    if tipo == 'object' or df[col].dtype.name == 'category':
        ejemplos = df[col].value_counts().head(2).to_dict()
    else:
        ejemplos = None

    resumen.append({
        'Variable': col,
        'Tipo': tipo,
        'Nulos': nulos,
        'Valores únicos': unicos,
        'Ejemplos / Top categorías': ejemplos
    })

resumen_df = pd.DataFrame(resumen)
resumen_df.head(20)
```

## Filtrando datos

Gracias a los operadores relacionales puedo filtrar columnas o filas.

Ejemplo: Quiero de mi dataframe de la columna **edad** a todos los estudiantes mayores o iguales a 12

```
df['edad'] == 12
```

El filtro anterior me da un columna con variables booleanas y lo puedo utilizar para formar un nuevo dataframe sin los valores False.

```
DF2=df[df['edad'] >= 12]
print(DF2)
```

Tambien puedo utilizar los operadores lógicos y concatenar mas de una consulta o filtro.

Ejemplo: Quiero a los mayores de 12 años y que sean de Mexico. A estos agregamelos en un nuevo dataframe.

```
data_ejemplo = df[(df['edad'] >= 12) & (df['pais'] == 'mx')]
print(data_ejemplo)
```

Tambien se puede utilizar la función query() de la siguiente manera.

```
print(df.query('edad > 12'))
```

```
##   Qué pasa cuando la logica es más compleja?
print(df.query("edad >= 12 and pais == 'mx'"))
```

La función query aumenta la legibilidad del código. Ejemplo:

```
##   Qué pasa cuando la logica es más compleja? Cuales son las diferencias

print(df.query( 'edad >= 12 and pais == "mx"'))
```

```
print( df[(df['edad'] >= 12) & (df['pais'] == 'mx')])
```

Muestran los mismos resultados pero el primero es mucho más legible que el segundo.

Ejemplo: Filtrame los que tengan en la columna de calificacion Q2 mayor o igual nota respecto a la columna Q1.

 1. DataFrames para append() – Unión vertical (más filas)
 Ejemplo: estudiantes nuevos

```
import pandas as pd

# Primer grupo de estudiantes
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Nombre': ['Ana', 'Luis', 'Carlos'],
    'Ciudad': ['Bogotá', 'Cali', 'Medellín']
})

# Segundo grupo de estudiantes
df2 = pd.DataFrame({
    'ID': [4, 5],
    'Nombre': ['Diana', 'Pedro'],
    'Ciudad': ['Barranquilla', 'Bogotá']
})


df_append = pd.concat([df1, df2], ignore_index=True)

print("🔹 DataFrame combinado con append/concat:")
print(df_append)
```

 2. DataFrames para merge() – Unión horizontal (por columnas compartidas)
 Ejemplo: información de estudiantes + notas

```

# Notas de los estudiantes
df_notas = pd.DataFrame({
    'ID': [1, 2, 4, 5],
    'Nota': [4.5, 3.8, 4.2, 3.9]
})

# Unión horizontal por ID
df_merge = pd.merge(df1, df_notas, on='ID', how='left')

print("🔹 DataFrame combinado con merge:")
print(df_merge)
```

```
## tratamiento de datos ausentes
import pandas as pd
import numpy as np

# Creamos un DataFrame de ejemplo
df = pd.DataFrame({
    'ID': [1, 2, 3, 4, 5],
    'Nombre': ['Ana', 'Luis', 'Carlos', 'Diana', 'Pedro'],
    'Ciudad': ['Bogotá', 'Cali', np.nan, 'Barranquilla', 'Bogotá'],
    'Edad': [23, np.nan, 30, 22, 27],
    'Nota': [4.5, 3.8, np.nan, 4.2, 3.9]
})

print(" DataFrame original con datos nulos:")
print(df)
```

```
df.isnull()
```

```
df.isnull().sum()
```

```
df_sin_nulos = df.dropna()

print(df_sin_nulos)
```

```
df_sin_columnas_nulas = df.dropna(axis=1)
print(df_sin_columnas_nulas)
```

```
df_filled = df.fillna(0)
print(df_filled)
```

```
df['Edad'] = df['Edad'].fillna(df['Edad'].mean())
print(df)
```

```
df.fillna(method='ffill', inplace=True)  # forward fill (rellena hacia abajo)
df.fillna(method='bfill', inplace=True)  # backward fill (hacia arriba)
```
